```python
import pandas as pd
import numpy as np
import ast
import os
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.metrics import (confusion_matrix, accuracy_score,
                             precision_score, recall_score, ConfusionMatrixDisplay)
from sklearn.utils.class_weight import compute_class_weight

import tensorflow as tf
from tensorflow.keras.preprocessing.image import load_img, img_to_array, ImageDataGenerator
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.models import Sequential, Model
from tensorflow.keras.layers import (Conv2D, MaxPooling2D, Flatten, Dense,
                                     Dropout, GlobalAveragePooling2D, BatchNormalization)
from tensorflow.keras.applications import ResNet50
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau
```

## 1. Load Dataset

```python
df = pd.read_csv("C:\Users\agilbert3\Downloads\archive (2).zip\full_df.csv")

# Convert string target -> list
df["target"] = df["target"].apply(ast.literal_eval)

print(df.head())
print("\nDataset shape:", df.shape)
```

## 2. Load Images

```python
IMG_SIZE = 224

TRAIN_DIR = "C:\Users\agilbert3\Downloads\archive (2).zip\ODIR-5K\ODIR-5K\Training Images"
TEST_DIR  = "C:\Users\agilbert3\Downloads\archive (2).zip\ODIR-5K\ODIR-5K\Testing Images"

def load_images(df):
    images, labels = [], []
    for _, row in df.iterrows():
        filename   = row["filename"]
        train_path = os.path.join(TRAIN_DIR, filename)
        test_path  = os.path.join(TEST_DIR,  filename)

        if os.path.exists(train_path):
            path = train_path
        elif os.path.exists(test_path):
            path = test_path
        else:
            print("Missing:", filename)
            continue

        img = load_img(path, target_size=(IMG_SIZE, IMG_SIZE))
        img = img_to_array(img) / 255.0
        images.append(img)
        labels.append(row["target"])

    return np.array(images), np.array(labels)

# Load all images then split
X_all, y_all = load_images(df)

# ── FIX: call train_test_split so X_train / X_test actually exist ────────────
X_train, X_test, y_train, y_test = train_test_split(
    X_all, y_all, test_size=0.2, random_state=42
)

print("X_train shape:", X_train.shape)
print("y_train shape:", y_train.shape)
print("X_test  shape:", X_test.shape)
print("y_test  shape:", y_test.shape)
print("Example label:", y_train[0])
print("Image  min/max:", X_train.min(), X_train.max())
```

## 3. Data Augmentation

```python
# ── FIX: datagen was defined but never used — now applied in model.fit() ─────
datagen = ImageDataGenerator(
    rotation_range=10,
    zoom_range=0.1,
    horizontal_flip=True
)
datagen.fit(X_train)
print("ImageDataGenerator configured and fitted.")
```

## 4. Class Weights (handle class imbalance)

```python
# ── FIX: use argmax only for class-weight calculation (one dominant label) ───
label_indices = np.argmax(y_train, axis=1)

class_weights_arr = compute_class_weight(
    class_weight="balanced",
    classes=np.unique(label_indices),
    y=label_indices
)
class_weights = dict(enumerate(class_weights_arr))
print("Class Weights:", class_weights)
```

## 5. Model 1 – ResNet50 (transfer learning backbone, no pretrained weights)

**Design rationale:** ResNet50 uses skip connections (residual blocks) that help
gradients flow through very deep networks, reducing vanishing-gradient problems.
We add Global Average Pooling to replace the original dense head, then two
fully-connected layers with dropout for regularisation before the 8-class
softmax output.


```python
base_model1 = ResNet50(
    weights=None,          # train from scratch (no pretrained weights)
    include_top=False,
    input_shape=(IMG_SIZE, IMG_SIZE, 3)
)

x = base_model1.output
x = GlobalAveragePooling2D()(x)
x = Dense(256, activation="relu")(x)
x = Dropout(0.5)(x)
x = Dense(128, activation="relu")(x)
outputs1 = Dense(8, activation="softmax")(x)

model1 = Model(inputs=base_model1.input, outputs=outputs1)

model1.compile(
    optimizer="adam",
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)
model1.summary()
```

### 5a. Train Model 1

```python
callbacks1 = [
    EarlyStopping(monitor="val_loss", patience=10, restore_best_weights=True),
    ReduceLROnPlateau(monitor="val_loss", factor=0.5, patience=5, min_lr=1e-6)
]

# ── FIX: use datagen.flow() so augmentation is actually applied ───────────────
history1 = model1.fit(
    datagen.flow(X_train, y_train, batch_size=32),
    validation_data=(X_test, y_test),
    epochs=100,            # Level 4 requirement: at least 100 epochs
    class_weight=class_weights,
    callbacks=callbacks1
)
```

## 6. Model 2 – Custom CNN (different architecture)

**Design rationale / justification for the change:**  
Model 1 (ResNet50) is very deep (~25M params) and relies on residual connections.
Model 2 is a lighter custom CNN with explicit BatchNormalization after every
Conv block.  BatchNorm normalises activations at each layer, often stabilising
training faster and acting as an implicit regulariser.  This makes it a
meaningfully different architecture choice — shallower, with batch-norm
regularisation instead of skip connections — allowing a fair architectural
comparison.


```python
model2 = Sequential([
    # Block 1
    Conv2D(32, (3, 3), activation="relu", padding="same", input_shape=(IMG_SIZE, IMG_SIZE, 3)),
    BatchNormalization(),
    MaxPooling2D(2, 2),

    # Block 2
    Conv2D(64, (3, 3), activation="relu", padding="same"),
    BatchNormalization(),
    MaxPooling2D(2, 2),

    # Block 3
    Conv2D(128, (3, 3), activation="relu", padding="same"),
    BatchNormalization(),
    MaxPooling2D(2, 2),

    # Block 4
    Conv2D(256, (3, 3), activation="relu", padding="same"),
    BatchNormalization(),
    MaxPooling2D(2, 2),

    Flatten(),
    Dense(256, activation="relu"),
    Dropout(0.5),
    Dense(8, activation="softmax")
])

model2.compile(
    optimizer="adam",
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)
model2.summary()
```

### 6a. Train Model 2

```python
callbacks2 = [
    EarlyStopping(monitor="val_loss", patience=10, restore_best_weights=True),
    ReduceLROnPlateau(monitor="val_loss", factor=0.5, patience=5, min_lr=1e-6)
]

history2 = model2.fit(
    datagen.flow(X_train, y_train, batch_size=32),
    validation_data=(X_test, y_test),
    epochs=100,            # Level 4 requirement: at least 100 epochs
    class_weight=class_weights,
    callbacks=callbacks2
)
```

## 7. Evaluate Both Models (confusion matrix, accuracy, precision, recall)

```python
CLASS_NAMES = ["N", "D", "G", "C", "A", "H", "M", "O"]

def evaluate_model(model, X_test, y_test, name):
    """Print confusion matrix, accuracy, precision, recall for a model."""
    y_pred_prob = model.predict(X_test)
    y_pred      = np.argmax(y_pred_prob, axis=1)
    y_true      = np.argmax(y_test,      axis=1)

    acc  = accuracy_score(y_true, y_pred)
    prec = precision_score(y_true, y_pred, average="weighted", zero_division=0)
    rec  = recall_score(y_true, y_pred, average="weighted", zero_division=0)

    print(f"\n{'='*50}")
    print(f"{name}")
    print(f"{'='*50}")
    print(f"Accuracy : {acc:.4f}")
    print(f"Precision: {prec:.4f}")
    print(f"Recall   : {rec:.4f}")

    cm = confusion_matrix(y_true, y_pred)
    disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=CLASS_NAMES)
    fig, ax = plt.subplots(figsize=(8, 8))
    disp.plot(ax=ax, colorbar=False, cmap="Blues")
    ax.set_title(f"Confusion Matrix – {name}")
    plt.tight_layout()
    plt.show()

    return {"accuracy": acc, "precision": prec, "recall": rec}

results1 = evaluate_model(model1, X_test, y_test, "Model 1: ResNet50")
results2 = evaluate_model(model2, X_test, y_test, "Model 2: Custom CNN")
```

## 8. Training Curves

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

for ax, history, label in zip(axes, [history1, history2],
                               ["Model 1: ResNet50", "Model 2: Custom CNN"]):
    ax.plot(history.history["accuracy"],     label="Train Accuracy")
    ax.plot(history.history["val_accuracy"], label="Val Accuracy")
    ax.set_title(f"Accuracy – {label}")
    ax.set_xlabel("Epoch")
    ax.set_ylabel("Accuracy")
    ax.legend()

plt.tight_layout()
plt.show()
```

## 9. Statistical Comparison of the Two Models

We use a **McNemar's test** (paired test on correct/incorrect predictions) and a
**95 % confidence interval** on the accuracy difference to determine whether one
model is statistically better than the other.


```python
from statsmodels.stats.contingency_tables import mcnemar
from scipy import stats

y_true_idx  = np.argmax(y_test, axis=1)
pred1_idx   = np.argmax(model1.predict(X_test), axis=1)
pred2_idx   = np.argmax(model2.predict(X_test), axis=1)

correct1 = (pred1_idx == y_true_idx)
correct2 = (pred2_idx == y_true_idx)

# ── Contingency table for McNemar ────────────────────────────────────────────
# [both wrong,  m1 wrong m2 right]
# [m1 right m2 wrong, both right ]
b = np.sum(~correct1 &  correct2)
c = np.sum( correct1 & ~correct2)
table = [[np.sum(~correct1 & ~correct2), b],
         [c, np.sum(correct1 & correct2)]]

result = mcnemar(table, exact=False, correction=True)
print("McNemar's test statistic:", result.statistic)
print("p-value                 :", result.pvalue)
if result.pvalue < 0.05:
    print("=> Statistically significant difference (p < 0.05)")
else:
    print("=> No statistically significant difference (p >= 0.05)")

# ── 95 % CI on accuracy difference ───────────────────────────────────────────
n   = len(y_true_idx)
acc_diff  = results1["accuracy"] - results2["accuracy"]
se_diff   = np.sqrt((results1["accuracy"]*(1-results1["accuracy"]) +
                     results2["accuracy"]*(1-results2["accuracy"])) / n)
z95       = 1.96
ci_low    = acc_diff - z95 * se_diff
ci_high   = acc_diff + z95 * se_diff

print(f"\nAccuracy difference (M1 - M2): {acc_diff:+.4f}")
print(f"95% Confidence Interval: [{ci_low:.4f}, {ci_high:.4f}]")
```

## 10. Real-Time Diagnosis Demo

Classify a single image in real time and display Type 1 / Type 2 error rates.


```python
LABEL_NAMES = {
    0: "Normal (N)",
    1: "Diabetes (D)",
    2: "Glaucoma (G)",
    3: "Cataract (C)",
    4: "AMD (A)",
    5: "Hypertension (H)",
    6: "Myopia (M)",
    7: "Other (O)"
}

def diagnose_image(image_path, model, class_names=LABEL_NAMES):
    """Load one image, run inference, print top prediction + error rates."""
    img = load_img(image_path, target_size=(IMG_SIZE, IMG_SIZE))
    img_arr = img_to_array(img) / 255.0
    img_arr = np.expand_dims(img_arr, axis=0)

    probs      = model.predict(img_arr)[0]
    pred_class = np.argmax(probs)
    confidence = probs[pred_class]

    print(f"Predicted class : {class_names[pred_class]}")
    print(f"Confidence      : {confidence:.2%}")
    print("\nAll class probabilities:")
    for i, p in enumerate(probs):
        print(f"  {class_names[i]:<20} {p:.4f}")

    # Type 1 error rate (false positive) – fraction predicted as this class that are wrong
    # Type 2 error rate (false negative) – fraction of true positives missed
    y_true_idx = np.argmax(y_test, axis=1)
    y_pred_idx = np.argmax(model.predict(X_test), axis=1)

    true_pos  = np.sum((y_pred_idx == pred_class) & (y_true_idx == pred_class))
    false_pos = np.sum((y_pred_idx == pred_class) & (y_true_idx != pred_class))
    false_neg = np.sum((y_pred_idx != pred_class) & (y_true_idx == pred_class))
    total_pred_pos = true_pos + false_pos
    total_actual_pos = true_pos + false_neg

    type1_rate = false_pos / total_pred_pos   if total_pred_pos   > 0 else 0.0
    type2_rate = false_neg / total_actual_pos if total_actual_pos > 0 else 0.0

    print(f"\nFor class '{class_names[pred_class]}':")
    print(f"  Type 1 error rate (false positive rate): {type1_rate:.2%}")
    print(f"  Type 2 error rate (false negative rate): {type2_rate:.2%}")

    plt.imshow(img)
    plt.title(f"Prediction: {class_names[pred_class]} ({confidence:.1%})")
    plt.axis("off")
    plt.show()

# ── Usage: replace with a real path to an eye fundus image ───────────────────
# diagnose_image("/path/to/test_image.jpg", model1)
# Example using a random image from the test set:
sample_idx  = np.random.randint(0, len(X_test))
sample_img  = X_test[sample_idx]
sample_true = LABEL_NAMES[np.argmax(y_test[sample_idx])]

print(f"True label: {sample_true}")

probs      = model1.predict(np.expand_dims(sample_img, axis=0))[0]
pred_class = np.argmax(probs)
confidence = probs[pred_class]

y_true_idx = np.argmax(y_test, axis=1)
y_pred_idx = np.argmax(model1.predict(X_test), axis=1)

true_pos  = np.sum((y_pred_idx == pred_class) & (y_true_idx == pred_class))
false_pos = np.sum((y_pred_idx == pred_class) & (y_true_idx != pred_class))
false_neg = np.sum((y_pred_idx != pred_class) & (y_true_idx == pred_class))
total_pred_pos   = true_pos + false_pos
total_actual_pos = true_pos + false_neg

type1_rate = false_pos / total_pred_pos   if total_pred_pos   > 0 else 0.0
type2_rate = false_neg / total_actual_pos if total_actual_pos > 0 else 0.0

print(f"Predicted class : {LABEL_NAMES[pred_class]} ({confidence:.2%})")
print(f"Type 1 error rate (false positive rate): {type1_rate:.2%}")
print(f"Type 2 error rate (false negative rate): {type2_rate:.2%}")

plt.imshow(sample_img)
plt.title(f"True: {sample_true} | Pred: {LABEL_NAMES[pred_class]} ({confidence:.1%})")
plt.axis("off")
plt.show()
```

## 11. GradCAM Heatmap

Visualise which regions of the fundus image the model focuses on when making
its prediction (Level 4 requirement).


```python
import tensorflow as tf

def make_gradcam_heatmap(img_array, model, last_conv_layer_name, pred_index=None):
    """Generate a GradCAM heatmap for the given image."""
    grad_model = tf.keras.models.Model(
        inputs  = model.input,
        outputs = [model.get_layer(last_conv_layer_name).output, model.output]
    )
    with tf.GradientTape() as tape:
        conv_outputs, predictions = grad_model(img_array)
        if pred_index is None:
            pred_index = tf.argmax(predictions[0])
        class_channel = predictions[:, pred_index]

    grads     = tape.gradient(class_channel, conv_outputs)
    pooled    = tf.reduce_mean(grads, axis=(0, 1, 2))
    conv_out  = conv_outputs[0]
    heatmap   = conv_out @ pooled[..., tf.newaxis]
    heatmap   = tf.squeeze(heatmap)
    heatmap   = tf.maximum(heatmap, 0) / (tf.math.reduce_max(heatmap) + 1e-8)
    return heatmap.numpy()

def display_gradcam(img_array_3d, model, last_conv_layer_name, title="GradCAM"):
    img_4d   = np.expand_dims(img_array_3d, axis=0)
    heatmap  = make_gradcam_heatmap(img_4d, model, last_conv_layer_name)

    heatmap_resized = tf.image.resize(
        heatmap[..., np.newaxis], (IMG_SIZE, IMG_SIZE)
    ).numpy().squeeze()

    # Overlay heatmap on original image
    import cv2
    heatmap_uint8 = np.uint8(255 * heatmap_resized)
    heatmap_color = cv2.applyColorMap(heatmap_uint8, cv2.COLORMAP_JET)
    heatmap_color = cv2.cvtColor(heatmap_color, cv2.COLOR_BGR2RGB) / 255.0
    overlay       = 0.6 * img_array_3d + 0.4 * heatmap_color

    fig, axes = plt.subplots(1, 3, figsize=(14, 5))
    axes[0].imshow(img_array_3d);            axes[0].set_title("Original");   axes[0].axis("off")
    axes[1].imshow(heatmap_resized, cmap="jet"); axes[1].set_title("Heatmap");    axes[1].axis("off")
    axes[2].imshow(np.clip(overlay, 0, 1));  axes[2].set_title("Overlay");    axes[2].axis("off")
    plt.suptitle(title)
    plt.tight_layout()
    plt.show()

# ResNet50: last conv layer is 'conv5_block3_out'
display_gradcam(X_test[sample_idx], model1,
                last_conv_layer_name="conv5_block3_out",
                title=f"GradCAM – Model 1 (ResNet50) | True: {sample_true}")

# Custom CNN: last conv layer is the 4th Conv2D
last_conv_name2 = [l.name for l in model2.layers if "conv2d" in l.name][-1]
display_gradcam(X_test[sample_idx], model2,
                last_conv_layer_name=last_conv_name2,
                title=f"GradCAM – Model 2 (Custom CNN) | True: {sample_true}")
```
